```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

# --- Load CSV file ---
df = pd.read_csv('../output_data/tmdb_clean_step1.csv')

attribute = "main_genre"

X = df.drop(columns=[attribute])
y = df[attribute]

# --- Stratified K-fold ---
skf = StratifiedKFold(n_splits=10, shuffle=True)

accuracies = []
f1_scores = []
conf_matrices = []

for fold, (train_idx, test_idx) in enumerate(skf.split(X, y), start=1):
    X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    # --- Scale inside each fold ---
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled  = scaler.transform(X_test)

    # --- Gaussian Naive Bayes ---
    nb = GaussianNB()
    nb.fit(X_train_scaled, y_train)

    # --- Predict ---
    y_pred = nb.predict(X_test_scaled)

    # --- Metrics ---
    acc = accuracy_score(y_test, y_pred)
    f1  = f1_score(y_test, y_pred, average="macro")
    cm  = confusion_matrix(y_test, y_pred)

    accuracies.append(acc)
    f1_scores.append(f1)
    conf_matrices.append(cm)

    print(f"\n--- Fold {fold} ---")
    print(f"Accuracy: {acc:.4f}")
    print(f"F1 (macro): {f1:.4f}")
    print("Confusion matrix:")
    print(cm)

# --- Summary ---
print("\n===============================")
print("Final Cross-Validation Results")
print("===============================")
print(f"Mean Accuracy: {np.mean(accuracies):.4f}")
print(f"Mean F1 Score (macro): {np.mean(f1_scores):.4f}")
print(f"Std Accuracy: {np.std(accuracies):.4f}")
print(f"Std F1 Score: {np.std(f1_scores):.4f}")

```

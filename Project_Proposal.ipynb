## 3. Implementation / Technical Merit

### Anticipated Challenges

- **Missing values:** Some movies may have missing values in features like maybe budget or runtime. We will need to decide remove those rows or keep it.  
- **Class sizes:** Some genres have many movies, while others have very fewer. Even with only Action, Comedy, and Drama, the groups may still be less than others. So we may add more genres if needed.
- **Many genres to one main genre:** A movie can have more than one genre. We must choose one main genre.

### Feature Selection

The dataset has many columns, but we do not need all of them.

### Columns we need to remove 

- **id**
- **homepage**
- **original_title**
- **tagline**
- **status**
- **keywords**
- **production_companies**
- **production_countries**
- **spoken_languages**
 
### We will keep simple and important numerical columns:

- **budget**
- **popularity**
- **vote_average**
- **vote_count**
- **runtime**
- **release_date** → we will change this to **release_year**
- **genres** → we will change to our three mian genres: **Action / Comedy / Drama**

### Why we choose these features?

- These columns contain numbers or simple values.
- They are more easier to work with compared to the whole large data.
- They help separate Action, Comedy, and Drama.

## 4. Potential Impact of the Results

### Why Are These Results Useful?

A genre prediction model can help us classify movies using basic movie data.

- It can **Organizing large movie catalogs** without someone manually choosing the genre.

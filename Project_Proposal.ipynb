Project Name: **Movie Genre Classification**

## Dataset Description

We will use a public movie dataset from Kaggle that contains information about thousands of movies, including metadata and genre labels.

**Dataset link:**  
https://www.kaggle.com/datasets/tmdb/tmdb-movie-metadata  

**Format:** CSV  
**Size:** Around 5,000–7,000 movies with more than 20 attributes.

### Dataset Contents

The dataset includes multiple useful attributes describing each movie, such as:

- **budget** 
- **popularity** 
- **vote_average**  
- **vote_count** 
- **runtime** 
- **genres** 
- **release_date** 
- **overview**  

### Class Information

The original dataset has many different genres for each movie, and some movies have more than one genre. So if we want to make our project more easier to work with, we will only keep three main genres:

**1.Action**

**2.Comedy**

**3.Drama**
